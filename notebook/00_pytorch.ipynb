```python
# Practice using basic square model
# A number is given to the model and it should square it


import torch

x = [1, 2, 3, 4, 5, 6, 7, 8, 9]
y = [1, 4, 9, 16, 25, 36, 49, 64, 81]
```

```python
torch_x = torch.tensor(x, dtype=torch.float32)
torch_y = torch.tensor(y, dtype=torch.float32)

u = torch_x * torch_x           # Square function

W = torch.tensor(0.01)
b = torch.tensor(0.00)


def forward(u, W, b):
    return W * u + b


def loss(pred, actual):
    error = pred - actual
    
    return (error * error).sum() / error.numel()


def gradient(u, pred, actual):
    error = pred - actual
    N = error.numel()
    
    d = (2 * (error)) / N
    dW = (d * u).sum()
    db = d.sum()
    
    return dW, db
```

```python
LEARNING_RATE = 0.0001

for epoch in range(5000):
    pred = forward(u, W, b)
    dW, db = gradient(u, pred, torch_y)
    W = W - LEARNING_RATE * dW
    b = b - LEARNING_RATE * db
    if epoch % 500 == 0:
        print(epoch, loss(pred, torch_y).item(), W.item(), b.item())
```

```python
def predict(x_new, W, b):
    x_new = torch.tensor(x_new, dtype=torch.float32)
    u_new = x_new * x_new
    return forward(u_new, W, b)

print(predict(-10.0, W, b))
print(predict(12.0, W, b))
print(predict(2.5, W, b))

print("Weight: ", W)
print("Bias: ", b)
```
